```python
%matplotlib inline
import matplotlib
matplotlib.use('module://matplotlib_inline.backend_inline')
import matplotlib.pyplot as plt
@w.interact(n=(1, 10), label='curve')
def f(n=4, label='curve'):
    print(label, n)
    plt.figure(figsize=(3, 2)); plt.plot([i ** n for i in range(5)]); plt.show()
```
